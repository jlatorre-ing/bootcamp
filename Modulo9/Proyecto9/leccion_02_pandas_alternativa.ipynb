```python
# ⚡ Lección 2 – Procesamiento de Datos (Versión Pandas)
### Proyecto: Retail Analytics Pipeline — RetailMax
**Módulo 9: Fundamentos de Big Data | Alkemy**

---
**Objetivo:** Procesar el dataset Fashion-MNIST usando pandas (alternativa eficiente a Spark para datasets pequeños).

## 1. Carga y Exploración del Dataset

import pandas as pd
import numpy as np
import os
from sklearn.model_selection import train_test_split

# Rutas del dataset
DATA_PATH = 'fashion_mnist'
TRAIN_CSV = os.path.join(DATA_PATH, 'fashion_train.csv')
TEST_CSV  = os.path.join(DATA_PATH, 'fashion_test.csv')

# Cargar datos
print('=== CARGANDO DATASET ===')
df_train = pd.read_csv(TRAIN_CSV)
df_test = pd.read_csv(TEST_CSV)

print(f'Train shape: {df_train.shape}')
print(f'Test shape : {df_test.shape}')
print(f'Columnas   : {list(df_train.columns[:5])}...')

# Ver primeras filas
print('\n=== PRIMERAS FILAS ===')
print(df_train.head(3))

## 2. Análisis Exploratorio

# Distribución de clases
print('\n=== DISTRIBUCIÓN DE CLASES ===')
print(df_train['label_name'].value_counts().sort_index())

# Estadísticas básicas
print('\n=== ESTADÍSTICAS DE PÍXELES ===')
pixel_cols = [col for col in df_train.columns if col.startswith('pixel_')]
print(f'Píxeles totales: {len(pixel_cols)}')
print(f'Rango de valores: [{df_train[pixel_cols].min().min()}, {df_train[pixel_cols].max().max()}]')

## 3. Preprocesamiento para ML

# Separar features y target
X_train = df_train[pixel_cols].values.astype(np.float32) / 255.0  # Normalizar
y_train = df_train['label'].values
X_test = df_test[pixel_cols].values.astype(np.float32) / 255.0
y_test = df_test['label'].values

print('\n=== DATOS PREPROCESADOS ===')
print(f'X_train shape: {X_train.shape}')
print(f'y_train shape: {y_train.shape}')
print(f'X_test shape : {X_test.shape}')
print(f'y_test shape : {y_test.shape}')

# Crear subconjunto para pruebas rápidas
X_train_small, _, y_train_small, _ = train_test_split(
    X_train, y_train, train_size=0.1, random_state=42, stratify=y_train
)

print(f'\nSubconjunto pequeño: {X_train_small.shape[0]} muestras')

## 4. Simulación de Procesamiento Distribuido

# Simular particionamiento (como en Spark)
n_partitions = 4
partition_size = len(X_train_small) // n_partitions

print('\n=== SIMULACIÓN DE PARTICIONAMIENTO ===')
for i in range(n_partitions):
    start_idx = i * partition_size
    end_idx = start_idx + partition_size if i < n_partitions - 1 else len(X_train_small)
    partition = X_train_small[start_idx:end_idx]
    print(f'Partición {i+1}: {partition.shape} muestras')

print('\n✅ Procesamiento completado exitosamente')
print('💡 Esta versión es mucho más rápida que Spark para datasets pequeños')
```
